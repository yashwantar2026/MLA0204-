```python

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report

# Create mobile dataset
data = {
    'RAM': [512, 1024, 2048, 4096, 512, 1024,
            2048, 4096, 3072, 6144, 8192, 2048],
    'Battery': [2000, 2500, 3000, 4000, 1800, 2600,
                3200, 4500, 3500, 5000, 5500, 2800],
    'Internal_Memory': [8, 16, 32, 64, 8, 16,
                        32, 128, 64, 128, 256, 32],
    'Camera': [5, 8, 12, 16, 5, 8,
               13, 20, 16, 24, 48, 12],
    'Screen_Size': [4.5, 5.0, 5.5, 6.0, 4.3, 5.0,
                    5.5, 6.2, 6.0, 6.5, 6.7, 5.4],
    'Price_Range': [0, 0, 1, 2, 0, 0,
                    1, 3, 2, 3, 3, 1]
}

df = pd.DataFrame(data)

# Select input and output
X = df[['RAM', 'Battery', 'Internal_Memory',
        'Camera', 'Screen_Size']]
y = df['Price_Range']

# Split dataset
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)

# Create Random Forest classifier
model = RandomForestClassifier(
    n_estimators=100, random_state=42
)

# Train model
model.fit(X_train, y_train)

# Predict test data
y_pred = model.predict(X_test)

# Evaluate model
print("Mobile Price Prediction")
print("Accuracy:",
      accuracy_score(y_test, y_pred) * 100, "%")

print("\nClassification Report:")
print(classification_report(y_test, y_pred,
                            zero_division=0))

# Predict price range of a new mobile
new_mobile = pd.DataFrame({
    'RAM': [4096],
    'Battery': [4000],
    'Internal_Memory': [64],
    'Camera': [16],
    'Screen_Size': [6.0]
})

prediction = model.predict(new_mobile)

price_ranges = {
    0: "Low Cost",
    1: "Medium Cost",
    2: "High Cost",
    3: "Very High Cost"
}

print("Predicted Mobile Price Range:",
      price_ranges[prediction[0]])
```
